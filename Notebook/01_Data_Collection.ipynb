```python
def calculate_iqr(data):

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_limit) |
        (data > upper_limit)
    ]

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", len(outliers))
```

```python
# 1. Import library
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 2. Marketing variables select karna
marketing_cols = [
    "TV_Impressions",
    "YouTube_Impressions",
    "Facebook_Impressions",
    "Instagram_Impressions",
    "Print_Readership",
    "Radio_Listenership",
    "Trade_Spend"
]

# 3. Data select karna
X = numeric_df1[marketing_cols].dropna()

# 4. VIF calculate karna
vif_data = pd.DataFrame()

vif_data["Feature"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

# 5. Result display karna
display(vif_data)

# 6. CSV save karna
vif_data.to_csv("Monday_VIF_Results.csv", index=False)

# 7. Text report save karna
with open("Monday_VIF_Analysis_Notes.txt", "w") as file:
    file.write("MONDAY VIF ANALYSIS\n")
    file.write("==================\n\n")
    file.write("Objective: Check multicollinearity among marketing variables.\n\n")
    file.write(vif_data.to_string(index=False))

print("Monday VIF results and notes saved successfully!")
```

```python
# Check constant columns
constant_cols = X.columns[X.nunique() <= 1].tolist()

# Check duplicate columns
duplicate_cols = X.columns[X.T.duplicated()].tolist()

# Check perfectly correlated columns
corr_matrix = X.corr()

perfect_corr = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
).stack()

perfect_corr = perfect_corr[perfect_corr.abs() >= 0.999999]

print("Constant Columns:", constant_cols)
print("Duplicate Columns:", duplicate_cols)
print("\nPerfectly Correlated Columns:")
display(perfect_corr)

# Save diagnostic results
with open("Monday_VIF_Diagnostic_Notes.txt", "w") as file:
    file.write(f"Constant Columns: {constant_cols}\n")
    file.write(f"Duplicate Columns: {duplicate_cols}\n")
    file.write("\nPerfect Correlations:\n")
    file.write(perfect_corr.to_string())

print("VIF diagnostic report saved!")
```

```python
# Standardize marketing variables
from sklearn.preprocessing import StandardScaler
from statsmodels.stats.outliers_influence import variance_inflation_factor

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Calculate VIF
vif_data = pd.DataFrame()

vif_data["Feature"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X_scaled, i)
    for i in range(X_scaled.shape[1])
]

display(vif_data)

# Save corrected VIF
vif_data.to_csv("Monday_VIF_Corrected.csv", index=False)

print("Corrected Monday VIF saved!")
```

```python
from statsmodels.stats.diagnostic import het_breuschpagan
import statsmodels.api as sm
import pandas as pd

# Monday regression
X_monday = sm.add_constant(df1["Instagram_Impressions"])
model_monday = sm.OLS(
    df1["YouTube_Impressions"], X_monday
).fit()

bp_monday = het_breuschpagan(
    model_monday.resid, model_monday.model.exog
)

# Saturday regression
X_saturday = sm.add_constant(df2["Instagram_Impressions"])
model_saturday = sm.OLS(
    df2["YouTube_Impressions"], X_saturday
).fit()

bp_saturday = het_breuschpagan(
    model_saturday.resid, model_saturday.model.exog
)

# Results table
bp_results = pd.DataFrame({
    "Dataset": ["Monday", "Saturday"],
    "LM Statistic": [bp_monday[0], bp_saturday[0]],
    "LM p-value": [bp_monday[1], bp_saturday[1]],
    "F Statistic": [bp_monday[2], bp_saturday[2]],
    "F p-value": [bp_monday[3], bp_saturday[3]]
})

# CSV save
bp_results.to_csv("Monday_Saturday_Breusch_Pagan.csv", index=False)

# TXT notes save
with open("Breusch_Pagan_Notes.txt", "w") as file:
    file.write("STEP 18: BREUSCH-PAGAN TEST\n\n")
    file.write("Purpose: Check heteroscedasticity.\n")
    file.write("Concept: Statistical Hypothesis Testing.\n\n")
    file.write(bp_results.to_string(index=False))
    file.write("\n\np-value < 0.05: Evidence of heteroscedasticity.")
    file.write("\np-value >= 0.05: Insufficient evidence of heteroscedasticity.")

print(bp_results)
print("\nCSV and Notes saved successfully!")
```

```python
import numpy as np
import pandas as pd

# Digital marketing channels
digital_cols = [
    "YouTube_Impressions",
    "Facebook_Impressions",
    "Instagram_Impressions"
]

results = []

for name, df in [("Monday", df1), ("Saturday", df2)]:

    # Standardize variables for fair comparison
    X = df[digital_cols]
    X_scaled = (X - X.mean()) / X.std()

    # Matrix rank and singular values
    rank = np.linalg.matrix_rank(X_scaled)
    singular_values = np.linalg.svd(
        X_scaled, compute_uv=False
    )

    condition_number = np.linalg.cond(X_scaled)

    results.append({
        "Dataset": name,
        "Number_of_Channels": len(digital_cols),
        "Matrix_Rank": rank,
        "Condition_Number": condition_number,
        "Smallest_Singular_Value": singular_values[-1]
    })

# Comparison table
multicollinearity_check = pd.DataFrame(results)

# CSV save
multicollinearity_check.to_csv(
    "Monday_Saturday_Final_Multicollinearity_Check.csv",
    index=False
)

# TXT notes save
with open("Final_Multicollinearity_Notes.txt", "w") as file:
    file.write("STEP 20: FINAL MULTICOLLINEARITY INVESTIGATION\n\n")
    file.write("Purpose: Check linear dependency among digital channels.\n")
    file.write("Concepts: Standardization, Matrix Rank, SVD, Condition Number.\n\n")
    file.write(multicollinearity_check.to_string(index=False))
    file.write("\n\nNote: No marketing channel has been deleted.")

print(multicollinearity_check)
print("\nCSV and TXT Notes saved successfully!")
```

## Step 23: Apply Adstock Transformation

Objective:
Apply geometric adstock to actual marketing channels.

Why:
Advertising effects may continue beyond the week in which advertising occurs.

Technique:
Geometric Adstock Transformation.

Formula:
Adstock_t = Spend_t + Alpha * Adstock_(t-1)

Initial Alpha:
0.5 (for learning and initial analysis)

```python
# Step 23: Actual Dataset Adstock Transformation

alpha = 0.5

media_columns = [
    "TV_Impressions",
    "YouTube_Impressions",
    "Facebook_Impressions",
    "Instagram_Impressions",
    "Print_Readership",
    "Radio_Listenership"
]

def calculate_adstock(series, alpha):
    result = []
    previous = 0

    for value in series:
        current = value + alpha * previous
        result.append(current)
        previous = current

    return result


# Monday Adstock
for col in media_columns:
    df1_sorted[col + "_Adstock"] = (
        df1_sorted.groupby(
            ["Geo", "Brand", "SKU"], sort=False
        )[col].transform(
            lambda x: calculate_adstock(x, alpha)
        )
    )


# Saturday Adstock
for col in media_columns:
    df2_sorted[col + "_Adstock"] = (
        df2_sorted.groupby(
            ["Geo", "Brand", "SKU"], sort=False
        )[col].transform(
            lambda x: calculate_adstock(x, alpha)
        )
    )


# Save transformed datasets
df1_sorted.to_csv("Monday_Adstock_Transformed.csv", index=False)
df2_sorted.to_csv("Saturday_Adstock_Transformed.csv", index=False)

print("Monday:", df1_sorted.shape)
print("Saturday:", df2_sorted.shape)
print("Adstock transformation and CSV files saved!")
```

```python
# Hill Saturation Function

def hill_saturation(x, alpha=1.0, gamma=None):

    x = np.asarray(x, dtype=float)

    if gamma is None:
        gamma = np.median(x)

    numerator = x ** alpha
    denominator = numerator + gamma ** alpha

    return numerator / denominator


print("Hill Saturation Function defined successfully!")
```

```python
# STEP 38: LINEAR REGRESSION MODEL TRAINING

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd
import joblib

# Create Monday model
monday_model = LinearRegression()

# Train Monday model
monday_model.fit(X_monday_train, y_monday_train)

# Predict Monday test data
monday_predictions = monday_model.predict(X_monday_test)


# Create Saturday model
saturday_model = LinearRegression()

# Train Saturday model
saturday_model.fit(X_saturday_train, y_saturday_train)

# Predict Saturday test data
saturday_predictions = saturday_model.predict(X_saturday_test)


# Evaluation function
def evaluate_model(actual, predicted, model_name):

    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)

    return {
        "Dataset": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2_Score": r2
    }


# Evaluate both models
monday_results = evaluate_model(
    y_monday_test, monday_predictions, "Monday"
)

saturday_results = evaluate_model(
    y_saturday_test, saturday_predictions, "Saturday"
)

# Create comparison table
model_comparison = pd.DataFrame([
    monday_results,
    saturday_results
])

display(model_comparison)

# Save evaluation results
model_comparison.to_csv(
    "Linear_Regression_Model_Comparison.csv",
    index=False
)

# Save predictions
pd.DataFrame({
    "Actual_Sales": y_monday_test.values,
    "Predicted_Sales": monday_predictions
}).to_csv("Monday_Linear_Regression_Predictions.csv", index=False)

pd.DataFrame({
    "Actual_Sales": y_saturday_test.values,
    "Predicted_Sales": saturday_predictions
}).to_csv("Saturday_Linear_Regression_Predictions.csv", index=False)

# Save trained models
joblib.dump(monday_model, "Monday_Linear_Regression_Model.pkl")
joblib.dump(saturday_model, "Saturday_Linear_Regression_Model.pkl")

print("Model results, predictions and trained models saved successfully!")
```

```python

# ==========================================================
# STEP 45: HC3 ROBUST REGRESSION
# BEFORE vs AFTER COMPARISON
# Marketing Mix Modeling Project
# ==========================================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import os

# ----------------------------------------------------------
# 1. CREATE OUTPUT FOLDER
# ----------------------------------------------------------

output_folder = "HC3_Analysis_Results"
os.makedirs(output_folder, exist_ok=True)

# ----------------------------------------------------------
# 2. PREPARE TRAINING DATA
# ----------------------------------------------------------

# Monday
monday_X = pd.DataFrame(X_monday_train).copy()
monday_y = np.asarray(y_monday_train).ravel()

# Saturday
saturday_X = pd.DataFrame(X_saturday_train).copy()
saturday_y = np.asarray(y_saturday_train).ravel()

# Add constant
monday_X = sm.add_constant(monday_X, has_constant="add")
saturday_X = sm.add_constant(saturday_X, has_constant="add")

# ----------------------------------------------------------
# 3. FIT ORIGINAL OLS REGRESSION MODELS
# ----------------------------------------------------------

monday_ols = sm.OLS(
    monday_y, monday_X
).fit()

saturday_ols = sm.OLS(
    saturday_y, saturday_X
).fit()

# ----------------------------------------------------------
# 4. APPLY HC3 ROBUST STANDARD ERRORS
# ----------------------------------------------------------

monday_hc3 = monday_ols.get_robustcov_results(
    cov_type="HC3"
)

saturday_hc3 = saturday_ols.get_robustcov_results(
    cov_type="HC3"
)

# ----------------------------------------------------------
# 5. CREATE BEFORE vs AFTER COMPARISON
# ----------------------------------------------------------

def create_comparison(original, robust, dataset):

    names = original.model.exog_names

    return pd.DataFrame({

        "Dataset": dataset,

        "Variable": names,

        "Original_Coefficient": np.asarray(original.params),

        "HC3_Coefficient": np.asarray(robust.params),

        "Original_Standard_Error": np.asarray(original.bse),

        "HC3_Standard_Error": np.asarray(robust.bse),

        "Original_P_Value": np.asarray(original.pvalues),

        "HC3_P_Value": np.asarray(robust.pvalues)

    })


monday_comparison = create_comparison(
    monday_ols, monday_hc3, "Monday"
)

saturday_comparison = create_comparison(
    saturday_ols, saturday_hc3, "Saturday"
)

comparison = pd.concat(
    [monday_comparison, saturday_comparison],
    ignore_index=True
)

print("BEFORE vs AFTER HC3 COMPARISON")

display(comparison.round(6))

# ----------------------------------------------------------
# 6. MODEL PERFORMANCE RESULTS
# ----------------------------------------------------------

hc3_results = pd.DataFrame({

    "Dataset": ["Monday", "Saturday"],

    "R_Squared": [
        monday_ols.rsquared,
        saturday_ols.rsquared
    ],

    "F_Statistic_HC3": [
        monday_hc3.fvalue,
        saturday_hc3.fvalue
    ],

    "F_P_Value_HC3": [
        monday_hc3.f_pvalue,
        saturday_hc3.f_pvalue
    ]

})

print("\nHC3 MODEL PERFORMANCE")

display(hc3_results.round(6))

# ----------------------------------------------------------
# 7. SAVE CSV FILES
# ----------------------------------------------------------

comparison.to_csv(
    os.path.join(
        output_folder,
        "HC3_Before_After_Comparison.csv"
    ),
    index=False
)

hc3_results.to_csv(
    os.path.join(
        output_folder,
        "HC3_Robust_Results.csv"
    ),
    index=False
)

# ----------------------------------------------------------
# 8. CREATE COMPARISON GRAPH
# ----------------------------------------------------------

plot_data = comparison[
    comparison["Variable"] != "const"
].copy()

plot_data["Label"] = (
    plot_data["Dataset"] + "_" + plot_data["Variable"].astype(str)
)

x = np.arange(len(plot_data))
width = 0.35

plt.figure(figsize=(13, 6))

plt.bar(
    x - width/2,
    plot_data["Original_Standard_Error"],
    width,
    label="Original Standard Error"
)

plt.bar(
    x + width/2,
    plot_data["HC3_Standard_Error"],
    width,
    label="HC3 Robust Standard Error"
)

plt.xticks(
    x,
    plot_data["Label"],
    rotation=45,
    ha="right"
)

plt.xlabel("Dataset and Variables")
plt.ylabel("Standard Error")

plt.title("Original vs HC3 Robust Standard Errors")

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        output_folder,
        "HC3_Before_After_Comparison.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ----------------------------------------------------------
# 9. SAVE ANALYSIS NOTES
# ----------------------------------------------------------

with open(
    os.path.join(output_folder, "HC3_Analysis_Notes.txt"),
    "w"
) as file:

    file.write("MARKETING MIX MODELING - HC3 ANALYSIS\n\n")

    file.write("BEFORE vs AFTER COMPARISON\n\n")

    file.write(comparison.to_string(index=False))

    file.write("\n\nMODEL PERFORMANCE\n\n")

    file.write(hc3_results.to_string(index=False))

    file.write("\n\nOBSERVATIONS:\n")

    file.write(
        "HC3 adjusts standard errors to account for "
        "heteroscedasticity in statistical inference.\n"
    )

    file.write(
        "HC3 does not automatically eliminate "
        "heteroscedasticity or change OLS coefficients.\n"
    )

# ----------------------------------------------------------
# 10. FINAL SUCCESS MESSAGE
# ----------------------------------------------------------

print("\n====================================")
print("HC3 ANALYSIS COMPLETED SUCCESSFULLY")
print("====================================")

print("1. HC3_Before_After_Comparison.csv")
print("2. HC3_Robust_Results.csv")
print("3. HC3_Before_After_Comparison.png")
print("4. HC3_Analysis_Notes.txt")

print("\nAll files saved in:", output_folder)
```
